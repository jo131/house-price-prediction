```python
import pandas as pd
import seaborn as sbn
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold
```

```python
data_train = pd.read_csv("train.csv")
data_test = pd.read_csv("test.csv")
Y = data_train[["SalePrice"]]
Y = Y.to_numpy()
print(data_train[["PoolArea"]].describe())
```

```python
data_train[["BedroomAbvGr"]].describe()
X_train, X_test = data_train[["BedroomAbvGr","TotRmsAbvGrd","LotArea","OverallQual","OverallCond","KitchenAbvGr","GarageArea","PoolArea"]], data_test[["BedroomAbvGr","TotRmsAbvGrd","LotArea","OverallQual","OverallCond","KitchenAbvGr","GarageArea","PoolArea"]]
bath_train, bath_test = data_train[["BsmtFullBath","BsmtHalfBath","FullBath","HalfBath"]],data_test[["BsmtFullBath","BsmtHalfBath","FullBath","HalfBath"]]
```

```python
X_train, X_test = X_train.to_numpy(), X_test.to_numpy()
bath_train, bath_test = bath_train.to_numpy(), bath_test.to_numpy()
bath_train = bath_train[:, 0] + 0.5*bath_train[:, 1] + bath_train[:, 2] + 0.5*bath_train[:, 3]
bath_test = bath_test[:, 0] + 0.5*bath_test[:, 1] + bath_test[:, 2] + 0.5*bath_test[:, 3]
```

```python
bath_train = bath_train.reshape(-1, 1)
bath_test = bath_test.reshape(-1, 1)

X_train = np.concatenate([X_train, bath_train], axis=1)
X_test = np.concatenate([X_test, bath_test], axis=1)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
E = []
regressor = LinearRegression()
kf = KFold(n_splits=5,shuffle=True,random_state=0)
for train, test in kf.split(X_train):
    x_train, y_train = X_train[train], Y[train]
    x_test, y_test = X_train[test], Y[test]

    regressor.fit(x_train,y_train)
    y_pred = regressor.predict(x_test)
    E.append(np.sum((y_pred-y_test)**2)/len(y_pred))
print(E)
print(np.var(Y))
```
